## Feature dan Target

```python
# visualisasi hubungan antara setiap fitur dan target
for kolom in dsbeton.iloc[:, :8]:
    plt.figure(figsize=(6, 4))
    plt.scatter(dsbeton[kolom], dsbeton.iloc[:, 8], alpha=0.6, color='b')
    plt.title(f"{kolom} vs Kekuatan Beton")
    plt.xlabel(kolom)
    plt.ylabel("Concrete Compressive Strength (MPa)")
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```

## Variabel

```python
x = dsbeton.iloc[:, :8].values
y = dsbeton.iloc[:, 8].values
```

Variabel X adalah variabel independen yang dimana dia variabel yang bisa mempengaruhi variabel lain, tetapi tidak terpengaruh oleh variabel lain, jadi independen pada data set ini terletak pada index 0 - 7. Dan dependen nya ada pada target di index ke 8 Concrete comperessive strenght (kekuatan beton megapascal).

## Modeling

```python
#Split Test
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.4, random_state=0)
```

### 40% untuk data uji x_test dan y_test nanti 60% sisanya digunakan untuk training x_train dan y_train. 

```python
#Training MODEL
from sklearn.ensemble import RandomForestRegressor
model = RandomForestRegressor()
```

```python
model.fit(x_train, y_train)
```

## Prediksi

```python
y_predik = model.predict(x_test)
```

```python
y_predik
```

### Evaluasi

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae = mean_absolute_error(y_test, y_predik)
mse = mean_squared_error(y_test, y_predik)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_predik)

print(f"Mean Absolute Error (MAE): {mae:.2f}")
print(f"Mean Squared Error (MSE): {mse:.2f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.2f}")
print(f"R-squared (R2 Score): {r2:.2f}")
```

R2 score 0.62 yang berarti model mampu menjelaskan sekitar 62% saja dari pola dari kekuatan tekanan beton.

MAE (8.15) dan RMSE (10.20) rata rata hasil predik tekanan beton meleset sekitar 8.15 MPa.

### Prediksi Data Baru

```python
databaru = [[540, 76, 100, 158, 3.0, 1040, 676, 28]]
prediksibaru = model.predict(databaru)
print("Prediksi: ", prediksibaru)
```

### Selanjutnya visualiasasi nilai asli vs hasil prediksi.

```python
plt.figure(figsize=(6, 6))
plt.scatter(y_test, y_predik, alpha=0.7, color='purple')
plt.xlabel("Nilai Asli (y_test)")
plt.ylabel("Hasil Prediksi (y_predik)")
plt.title("Asli vs Prediksi")
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'k--', lw=2) # Garis diagonal ideal
plt.grid(True)
plt.show()
```

### Interpretasi
- Ttitik tidak berada rapat pada garis hitam, melainkan menyebar atau mengelompok membalut garis tersebut.

- Jarak vertikal antara titik ungu ke garis hitam menunjukkan melesetnya prediksi (penyebab MAE 8.15 dan RMSE 10.20 yang di dapatkan sebelumnya).

- Pada beton kekuatan rendah (Nilai Asli < 20 MPa): Banyak titik berada di atas garis hitam. yang dimana model cenderung menebak kekuatan beton lebih tinggi dari aslinya (overestimation).

- Pada beton kekuatan tinggi (Nilai Asli > 60 MPa): Banyak titik berada di bawah garis hitam. Artinya, model cenderung menebak kekuatan beton lebih rendah dari aslinya (underestimation).

### Kesimpulan

Grafik diatas memperjelas skor R2 model Linear Regression berada di angka 0.62.  karena hubungan material beton sebenarny adalah non-linier, model garis lurus (Linear Regression) sedikit kesulitan memprediksi nilai yang sangat rendah atau sangat tinggi secara presisi.

Jadi kesimpulannya kita bisa menggunakan regresi linnear tetapi hasilnya sedikit meleset atau tidak akurat. Jadi untuk mendapatkan hasil yang bagus dan akurat bisa mencoba dengan algoritma lain yang lebih cocok. Seperti Random Forest
